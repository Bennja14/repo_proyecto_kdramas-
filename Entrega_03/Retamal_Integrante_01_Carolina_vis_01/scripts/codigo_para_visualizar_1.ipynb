```python
df = pd.read_csv("kdramas_entrega3ggcolab.csv")
```

```python
df.head()
```

```python
df.columns = df.columns.str.strip()
```

```python
kdramas_por_ano = df.groupby("release_year").size().reset_index(name="cantidad")
```

```python
maximo = kdramas_por_ano["cantidad"].max()
```

```python
grafico = alt.Chart(kdramas_por_ano).mark_bar().encode(
    x=alt.X("release_year:O", title="Año"),
    y=alt.Y("cantidad:Q", title="Cantidad de K-Dramas"),

    color=alt.when(
        alt.datum.cantidad == maximo
    ).then(
        alt.value("purple")
    ).otherwise(
        alt.value("skyblue")
    ),

    tooltip=["release_year","cantidad"]

).properties(
    title="K-Dramas por año entre 2016 y 2025",
    width=600,
    height=400
)

grafico
```
