```python
df = pd.read_csv("kdramas2016real.csv", sep=';')
```

```python
df.columns = df.columns.str.strip()
```

```python
plataformas = df.melt(
    id_vars=["release_year"],

    value_vars=[
        "netflix",
        "viki",
        "prime_video",
        "disney_plus"
    ],

    var_name="plataforma",
    value_name="disponible"
)
```

```python
plataformas = plataformas[plataformas["disponible"] == 'yes']
```

```python
plataformas_por_ano = plataformas.groupby(
    ["release_year", "plataforma"]
).size().reset_index(name="cantidad")
```

```python
plataformas_por_ano["plataforma"] = plataformas_por_ano["plataforma"].replace({
    "netflix": "Netflix",
    "viki": "Viki",
    "prime_video": "Prime Video",
    "disney_plus": "Disney+"
})
plataformas["plataforma"] = plataformas["plataforma"].replace({
    "netflix": "Netflix",
    "viki": "Viki",
    "prime_video": "Prime Video",
    "disney_plus": "Disney+"
})
```

```python
df.head()
```

```python
grafico = alt.Chart(plataformas).mark_bar().encode(

    x=alt.X(
        'release_year:O',
        title='Año'
    ),

    y=alt.Y(
        'count():Q',
        title='Cantidad de K-Dramas'
    ),

    color=alt.Color(
    'plataforma:N',

    scale=alt.Scale(
        domain=['Netflix', 'Viki', 'Prime Video', 'Disney+'],

        range=[
            'red',
            '#00BFFF',
            '#9370DB',
            '#1E90FF'
        ]
    )
),

    tooltip=[
        alt.Tooltip("release_year:O", title="Año"),
        alt.Tooltip("plataforma:N", title="Plataforma"),
        alt.Tooltip("count():Q", title="Cantidad")
    ]

).properties(
    title='K-Dramas por plataforma y año (2016-2025)',
    width=700,
    height=400
)

grafico
```

```python
# Exportar la visualización:

grafico.save("vis_01.html")
```
