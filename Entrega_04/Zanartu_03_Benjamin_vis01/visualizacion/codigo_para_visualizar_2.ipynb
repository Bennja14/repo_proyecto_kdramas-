```python
import pandas as pd
import altair as alt
```

```python
df = pd.read_csv("kdramas_database_limpia.csv", delimiter=",", encoding="UTF-8")
```

```python
df.sort_values("ratings", ascending=False).head(10)
```

```python
top_kdramas = df.sort_values(by="ratings", ascending=False)
```

```python
top_100 = top_kdramas.head(100)
```

```python
conteo_plataformas = {
    "Netflix": (top_100["netflix"] == "yes").sum(),
    "Viki": (top_100["viki"] == "yes").sum(),
    "Prime Video": (top_100["prime_video"] == "yes").sum(),
    "Disney+": (top_100["disney_plus"] == "yes").sum()
}
```

```python
conteo_df = pd.DataFrame({
    "Platform": conteo_plataformas.keys(),
    "Cantidad": conteo_plataformas.values()
})
```

```python
alt.Chart(conteo_df).mark_bar().encode(
    x=alt.X('Platform', sort=None, title='Plataforma'),
    y=alt.Y('Cantidad', title='Número de KDramas'),
    color=alt.Color('Platform', scale=alt.Scale(
        domain=['Netflix', 'Viki', 'Prime Video', 'Disney+'],
        range=['red', 'green', 'lightblue', 'blue']
    ), legend=alt.Legend(title='Plataforma')),
    tooltip=['Platform', 'Cantidad']
).properties(
    title='Número de KDramas por Plataforma en el Top 100'
)
```

```python
chart_to_save = alt.Chart(conteo_df).mark_bar().encode(
    x=alt.X('Platform', sort=None, title='Plataforma'),
    y=alt.Y('Cantidad', title='Número de KDramas'),
    color=alt.Color('Platform', scale=alt.Scale(
        domain=['Netflix', 'Viki', 'Prime Video', 'Disney+'],
        range=['red', 'green', 'lightblue', 'blue']
    ), legend=alt.Legend(title='Plataforma')),
    tooltip=['Platform', 'Cantidad']
).properties(
    title='Número de KDramas por Plataforma en el Top 100'
)

chart_to_save.save('numero_kdramas_por_plataforma.html')
```
