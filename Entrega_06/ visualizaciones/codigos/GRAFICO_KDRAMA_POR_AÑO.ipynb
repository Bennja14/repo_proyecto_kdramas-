```python
import pandas as pd

# Replace 'your_database.csv' with the actual path to your database file
df = pd.read_csv('KDRAMAS BASE CUATRO MIL.csv')

# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
import altair as alt
```

```python
df_cleaned = df.copy()
df_cleaned.columns = df_cleaned.columns.str.strip() # Clean column names by stripping whitespace
df_cleaned['Year'] = pd.to_numeric(df_cleaned['Year'], errors='coerce')
df_cleaned = df_cleaned.dropna(subset=['Year'])
df_cleaned['Year'] = df_cleaned['Year'].astype(int)
```

```python
import altair as alt

alt.renderers.enable('default', embed_options={'actions': False})

df_cleaned = df.copy()
df_cleaned.columns = df_cleaned.columns.str.strip() # Clean column names by stripping whitespace
df_cleaned['Year'] = pd.to_numeric(df_cleaned['Year'], errors='coerce')
df_cleaned = df_cleaned.dropna(subset=['Year'])
df_cleaned['Year'] = df_cleaned['Year'].astype(int)

kdramas_per_year = df_cleaned.groupby('Year').size().reset_index(name='Number of KDramas')

chart = alt.Chart(kdramas_per_year).mark_bar().encode(
    x=alt.X('Year:O', title='Año'), # 'O' for ordinal to treat years as categories
    y=alt.Y('Number of KDramas', title='Número de Kdramas'),
    tooltip=['Year', 'Number of KDramas']
).properties(
    title='Número de Kdramas por año'
)

chart.display()
```

```python
chart.save('kdramas_per_year_chart.html', embed_options={'actions': False, 'vega': {'width': 800}, 'vega-lite': {'width': 800}})
```
