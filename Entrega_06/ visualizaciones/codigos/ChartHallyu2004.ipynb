```python
import pandas as pd
import altair as alt
```

```python
# Cargar CSV
df = pd.read_csv("hallyuchilecorregido.csv", sep=';')
```

```python
# Cambiar nombre de columna de fecha
df = df.rename(columns={df.columns[0]: "Fecha"})
```

```python
# Convertir fecha
df['Fecha'] = pd.to_datetime(df['Fecha'], format='%d-%m-%Y')
```

```python
# Convertir df a formato largo (long format)
df_long = df.melt(id_vars=['Fecha'], var_name='Termino', value_name='Indice')

# Convertir la columna 'Indice' a numérica, manejando errores
df_long['Indice'] = pd.to_numeric(df_long['Indice'], errors='coerce')

# Eliminar filas con valores NaN en 'Indice' si los hubiera
df_long = df_long.dropna(subset=['Indice'])

display(df_long.head())
```

```python
# Obtener el punto más alto de cada término
maximos = df_long.loc[
    df_long.groupby("Termino")["Indice"].idxmax()
]
```

```python
lineas = alt.Chart(df_long).mark_line(strokeWidth=2).encode(
    x=alt.X("Fecha:T", title="Año"),
    y=alt.Y("Indice:Q", title="Índice de búsqueda"),
    color=alt.Color("Termino:N", title="Término")
)
```

```python
puntos_maximos = alt.Chart(maximos).mark_circle(size=120).encode(
    x="Fecha:T",
    y="Indice:Q",
    color="Termino:N",
    tooltip=[
        alt.Tooltip("Termino:N", title="Término"),
        alt.Tooltip("Fecha:T", title="Fecha del máximo"),
        alt.Tooltip("Indice:Q", title="Índice máximo")
    ]
)
```

```python
chart = (lineas + puntos_maximos).properties(
    width=800,
    height=450,
    title="Interés por términos asociados al Hallyu en Chile"
)

chart
```

```python
# Exportar la visualización:

chart.save("hallyuchilecorregido.html")
```
